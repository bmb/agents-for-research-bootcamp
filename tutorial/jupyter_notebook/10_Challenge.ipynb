## Research question

Consider this research question: **Is the relationship between penguin flipper length and body mass consistent across species?**

In order to answer that question, you have at your disposal the [Palmer Penguins dataset](https://allisonhorst.github.io/palmerpenguins/articles/intro.html).

The data contains size measurements for three penguin species observed on three islands in the Palmer Archipelago, Antarctica.

These data were collected from 2007 - 2009 by Dr. Kristen Gorman with the Palmer Station Long Term Ecological Research Program, part of the US Long Term Ecological Research Network. The data were imported directly from the Environmental Data Initiative (EDI) Data Portal, and are available for use by CC0 license (“No Rights Reserved”) in accordance with the Palmer Station Data Policy.

Overall, your challenge is to

- Download the data
- Check missing values
- Plot the overall relationship
- Investigate differences between species using charts and numerical comparisons
- Make sure your analysis is reproducible



# Questions to get you started:

1. Which penguin species is heaviest?
2. Are penguins with longer flippers heavier?
3. Do penguins with deeper bills weigh more?
4. How does the male–female difference in body mass vary by species
5. Are penguins on Biscoe Island heavier?
6. Does the way we handle missing data change our conclusion?

## Possible steps

1. Create a new project folder, and add an AGENTS.md.
2. State the question (and what would count as a convincing answer).
3. Let the agent conduct an initial investigation.
4. Inspect a result and challenge an assumption.
5. Have the agent choose and run a follow-up check.
6. Verify one claim independently and write a short conclusion.
7. Save the workflow so it is easily reproducible.
