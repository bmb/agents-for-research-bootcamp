# A short introduction to OpenClaw

<div style="text-align: center;">
  <img src="images/openclaw_logo.png" style="width: 900px; height: auto;">
</div>

OpenClaw is an open-source AI assistant you run on your own computer or server. You can talk to it through familiar messaging apps such as WhatsApp, Telegram, Slack, or Discord, and connect it to tools that carry out tasks.

It brings together:
- An AI model — a hosted model or a local one.
- Tools and integrations — to work with files, code, and connected services.
- An always-running gateway — to connect your chats to the assistant and manage sessions.

For example, you could message it to research a topic, prepare a document, or work on a coding task, depending on the tools you’ve enabled.

Your assistant’s state and credentials stay on your hardware, although requests still go to whichever model provider and messaging platforms you configure.

## Installation

In a terminal, you can install OpenClaw with:

```bash
curl -fsSL https://openclaw.ai/install.sh | bash
```

<div style="text-align: center;">
  <img src="images/openclaw_setup.png" style="width: 600px; height: auto;">
</div>

The installer starts the guided onboarding wizard automatically. Choose `Quick start` to reuse detected AI access and open the dashboard, or `Custom setup` for the full guided flow. For this workshop, please use `Quick start` and follow the wizard.

Create one agent.

<div style="text-align: center;">
  <img src="images/openclaw_setup_custom_agent.png" style="width: 400px; height: auto;">
</div>

You can always return later with `openclaw configure` for additional settings, or with `openclaw onboard --classic` to open the classic step-by-step wizard instead.

### Local / NIM endpoint

When asked for a Model provider, select "More...", then pick **vLLM (Local/self-hosted OpenAI-compatible)**

<div style="text-align: center;">
  <img src="images/openclaw_setup_model_provider.png" style="width: 600px; height: auto;">
</div>

Then configure vLLM with the following setings:

- base url: `http://127.0.0.1:8000/v1`
- vLLM API key: anything you want, but must not be empty (even if not used by the vLLM NIM inference endpoint)
- vLLM model: `nvidia/nemotron-3.5-lightning`

<div style="text-align: center;">
  <img src="images/openclaw_setup_vllm_config.png" style="width: 600px; height: auto;">
</div>

### Finalize setup

Once the steps are done, OpenClaw will start the Gateway in the foreground, and you can access the web dashboard where you can add plugins, channels, skills, automations,... and start chatting with your AI assistant.

If you installed OpenClaw on your own local machine, then the dashboard would open automatically.

<div style="text-align: center;">
  <img src="images/openclaw_gateway_on.png" style="width: 900px; height: auto;">

### Starting the gateway manually

The gateway should start in the foreground after the install script. If not, run 

```bash
openclaw gateway run
```

### ... or install the gateway as a service (optional)

```bash
chmod 0700 ~/.config/    # needed on Brev instance, probably not on your local machine
openclaw gateway install
openclaw gateway start
```

### Adding web search skill

For this workshop, we will use Tavily as the web search provider. 

💡 **Tip**: You can create a free API key [here](https://www.tavily.com/), using Researcher as your persona.

Run the configuration tool:
```bash
openclaw configure
``` 

1. Select `Local (this machine)` then `Web tools`, and `Enable the web_search tool`.
2. Select the `Tavily` search provider and add the API key.
3. Use `Download from npm` to install the Tavily plugin.
4. Enable the `web-fetch` tool.
5. Select `Done`
6. ⚠️ **Restart the gateway** (see below).

<div style="text-align: center;">
  <img src="images/openclaw_configure_websearch.png" style="width: 900px; height: auto;">
</div>

### Adding NVIDIA Skills

You might need to reinstall the skills and add OpenClaw as one of the agent type (see <a href="09_Using_skills.ipynb">Agent Skills notebook</a>). In a terminal, run:

```bash
npx skills add nvidia/skills
```

### Restart the Gateway

- If running on the foreground, use `CTRL + c`, then `openclaw gateway run` in the terminal where the gateway is running
- If running as a service: `openclaw gateway restart`

#### Forwarding port on Brev

Please forward the instance port 18789 to a port (eg 18789 or any other free port) on your local machine using the Brev CLI (see <a href="../start_here.ipynb">Home Page</a>):

```bash
brev login
brev port-forward BREV_INSTANCE_NAME -p 18789:18789   # Replace BREV_INSTANCE_NAME with the actual name of the instance.
```

# Enforcing boundaries for safer agents: OpenShell and NemoClaw

OpenClaw has all access to your computer. NemoClaw wraps OpenClaw into OpenShell, a safe execution runtinme for agents, acting like a secure sandbox where policies enforce what the agent has access to (network, tools, files). The default policy is to deny everything.

NVIDIA NemoClaw is a collection of open blueprints for building autonomous agents: domain-specialized, always-on AI systems that reason, plan, and act across real-world workflows.  

Each blueprint includes NVIDIA Agent Toolkit components such as NVIDIA Nemotron™ and other frontier models, NVIDIA NeMo™ for specialization and optimization, and NVIDIA OpenShell™ for runtime policy controls.

**Note:** You can use NemoClaw with different agent harnesses, including OpenClaw, Hermes, LangChain, etc...
