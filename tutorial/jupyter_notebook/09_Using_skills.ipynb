## What are skills?

Reusable instructions that teach an agent how to perform a specific task.

<div style="text-align: center;">
  <img src="images/skillmd.png" style="width: 900px; height: auto;">
</div>

They are detailed in a folder containing task guidance and supporting resources that an agent loads when relevant:

<div style="text-align: center;">
  <img src="images/skills_ref.png" style="width: 900px; height: auto;">
</div>

Skills enable:
- Consistency: repeat a proven workflow.
- Reuse: share expertise across tasks.
- Efficiency: load details only when needed.

## NVIDIA Skills

NVIDIA continuously releases new skills teaching agents how to use NVIDIA software optimally: Physical AI and robotics workflows, simulation, CUDA-X libraries, RAG and AI Blueprints, and platform tools. 

This [NVIDIA Agent Skills](https://github.com/NVIDIA/skills) repository is a catalog: skills are maintained in their respective product repos, and mirrored here daily via an automated sync pipeline. Skills are being added continuously, so check back for updates.


### Installation

In a new terminal, if not done already, please install `npx`
```bash
curl -o- https://raw.githubusercontent.com/nvm-sh/nvm/v0.40.1/install.sh | bash
# you might need to exit and re-open a terminal for that
nvm install --lts
```

Then install NVIDIA skills with the default skills CLI flow using this command in the terminal:

```bash
npx skills add nvidia/skills
```

Once installed, the skills will be available in OpenCode. Use `/skills` to list and use them.

### Steps

#### Create the environment

1. Open a new terminal and create a new folder `test-skills`
2. Install the NVIDIA Skills, either globally or for the current project. OpenCode will find those automatically.
3. Create an AGENTS.md file, using `/init` inside `opencode` or copying from previous lab

#### Step 1: create a baseline.


Use this prompt:

```
Build and run a baseline for the delivery-routing problem below. Use Google OR-Tools Routing Solver in Python.

**Problem**

Two vans deliver boxes from one warehouse. Each van can carry at most 10 boxes. Every customer must be visited exactly once, and every used van must start and finish at the warehouse. Minimize total route distance. There are no time windows or service times.

| Location | x | y | Boxes |
|---|---:|---:|---:|
| Warehouse | 0 | 0 | 0 |
| A | 2 | 1 | 4 |
| B | 3 | 2 | 3 |
| C | 2 | 4 | 3 |
| D | -2 | 1 | 4 |
| E | -3 | 2 | 3 |
| F | -2 | 4 | 3 |

Use Euclidean distances in arbitrary units.
```


<div style="text-align: center;">

#### Step 2: using the cuOpt skills

1. Use the following prompt to install cuOpt:

```
Use the cuopt-install skill to inspect my environment for running cuOpt’s Python routing API. Check the operating system, GPU, driver, CUDA compatibility, and existing Python packages. Explain what is ready and what needs setup before making changes.
```

<div style="text-align: center;">

Instruct the agent to install cuOpt when it ask what are the next steps.

2. Use the cuopt-routing-api-python skill to solve the vehicle-routing demo in Python.

```
Convert the delivery-routing problem to cuOpt's  DataModel format, solve the problem with cuopt usint the cuopt-routing-api-python skill, and verify cuopt produces the same optimal solution (20.72 total distance).
```

If cuOpt installation fails, make sure you instruct the agent to install it in a dedicated virtual environment.
