# Turn Agents Into Research Assistant
---

As Generative AI evolves, research institutions are transitioning from passive chat interfaces to autonomous, goal-driven AI agents capable of reasoning, planning, and executing complex tasks. This hands-on workshop introduces participants to the fundamentals of Agentic AI using NVIDIA’s open-source solutions.

Designed for research software engineers, data specialists, and IT infrastructure professionals, this session moves beyond traditional prompt engineering to explore how autonomous agents can significantly enhance research productivity and data workflows.

You will learn how to design, implement, and orchestrate specialised agents or “claws” that function as persistent, background research assistants. The session will cover the following topics:

– Construct a functional autonomous agent tailored for domain-specific research applications.  
– Master reusable patterns.
– Mitigate technical debt by leveraging industry-standard frameworks

The local execution models provide flexible frameworks and you will leave equipped with the practical knowledge needed to build secure, domain-specific AI assistants.

This workshop bridges the gap between state-of-the-art AI tooling and daily research data infrastructure, empowering institutions to shape the future of automated, reproducible research.

## Learning Objectives

1. Understand the capabilities of NVIDIA Inference Microservices (NIMs) and how to use NIM endpoints for accelerated AI inference.
2. Understand the Model Context Protocol (MCP) and how it enables standardized communication tools and data sources.
3. Create MCP servers using the high-level Server API with tool definitions and input schemas.
4. Leveraging NIMs and MCP inside a coding agent like OpenCode
5. Developing an AI agent for a research workflows
6. Understanding agents skills and how to use them
7. Apply learned concepts in a challenge exercise to build a complete agentic application.

## Brev port forwarding

In some labs, you will need to forward some ports to your local machine. This can be done from the Brev CLI by following those steps:

1. Install the CLI:

On Linux/Windows (WSL):
```bash
bash -c "$(curl -fsSL https://raw.githubusercontent.com/brevdev/brev-cli/main/bin/install-latest.sh)"
```
On Windows:
```bash
brew install brevdev/homebrew-brev/brev
```

3. Login from the CLI:

```bash
brev login
```

4. Forward the port:

```
brev port-forward $instance_name -p $local:$remote
```

Note: The instructions are also displayed on the Brev instance page.
